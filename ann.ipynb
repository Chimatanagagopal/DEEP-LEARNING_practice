```python
import keras #library for neural network
import pandas as pd #loading data in table form
import seaborn as sns #visualisation
import matplotlib.pyplot as plt #visualisation
import numpy as np # linear algebra
from sklearn.preprocessing import normalize # machine learning algorithm library
```

```python
data=pd.read_csv("/content/iris.csv")
data.head()
```

```python
data["species"].unique()
```

```python
data.loc[data["species"]=="setosa","species"]=0
data.loc[data["species"]=="versicolor","species"]=1
data.loc[data["species"]=="virginica","species"]=2
```

```python
data=data.iloc[np.random.permutation(len(data))]
print(data)
```

```python
X=data.iloc[:,:4].values
y=data.iloc[:,4].values
```

```python
X_normalized = normalize(X,axis=0)
print("Examples of X_normalised\n",X_normalized[:3])
```

```python
total_length=len(data)
train_length=int(0.8*total_length)
test_length=int(0.2*total_length)

X_train=X_normalized[:train_length]
X_test=X_normalized[train_length:]
y_train=y[:train_length]
y_test=y[train_length:]
```

```python
#Neural network module
from keras.models import Sequential
from keras.layers import Dense,Activation
# from keras.utils import np_utils
import tensorflow as tf
```

```python
y_train=tf.keras.utils.to_categorical (y_train,num_classes=3)
y_test=tf.keras.utils.to_categorical (y_test,num_classes=3)
```

```python
model = Sequential()
model.add(Dense(32,input_dim=4,activation="relu"))
model.add(Dense(64,activation="relu"))
model.add(Dense(128,activation="relu"))
model.add(Dense(256,activation="relu"))
model.add(Dense(3, activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',# error
              optimizer='adam',# gradient descent
              metrics=['accuracy'] # accuracy
              )
```

```python
model.fit(X_train,y_train,
          validation_data=(X_test,y_test),
          batch_size=30,
          epochs=30)
```

```python
prediction=model.predict(X_test)
length=len(prediction)
y_label=np.argmax(y_test,axis=1)
predict_label=np.argmax(prediction,axis=1)
accuracy=np.sum(y_label==predict_label)/length * 100
print("Accuracy of the dataset",accuracy )
```

```python
correct = np.sum(y_label == predict_label)
wrong = np.sum(y_label != predict_label)

print("Total test data:", length)
print("Correct predictions:", correct)
print("Wrong predictions:", wrong)


print("Accuracy:", correct / length * 100)
```
