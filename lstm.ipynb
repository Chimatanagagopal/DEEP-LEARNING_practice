```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense


# Data
X = np.array([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6],
    [5, 6, 7]
])

y = np.array([4, 5, 6, 7, 8])


# Reshape for LSTM
X = X.reshape(5, 3, 1)


# Create model
model = Sequential([
    LSTM(10, input_shape=(3, 1)),
    Dense(1)
])


# Compile
model.compile(
    optimizer="adam",
    loss="mse"
)


# Train
model.fit(X, y, epochs=100, verbose=0)


# Prediction
prediction = model.predict(
    np.array([[[6], [7], [8]]])
)

print(prediction)
```
